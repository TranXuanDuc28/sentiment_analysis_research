### 8. Trực quan hóa kết quả (Confusion Matrix & t-SNE)
Hiển thị các biểu đồ đã được sinh ra trong quá trình thực nghiệm.

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import glob
import os

plots = glob.glob('results/plots/*.png')
for plot_path in sorted(plots):
    plt.figure(figsize=(10, 7))
    img = mpimg.imread(plot_path)
    plt.imshow(img)
    plt.axis('off')
    plt.title(os.path.basename(plot_path))
    plt.show()
```
